* EX 1.7 (2 points) = Make a template/script for the calculation of pH of a strong or a weak acid HA.
    - https://chem.libretexts.org/Bookshelves/Introductory_Chemistry &rarr; Search: pH of weak acid
    - Define two functions: pH_of_strong_acid(c), pH_of_weak_acid(c, pKa).
    - Print pH of a strong acid with c = 0.1, 0.2, 0.3, 0.4.

```python
# =============================================================================
# Import modules
# =============================================================================
from math import log, sqrt
```

```python
# =============================================================================
# Input variables
# =============================================================================
c = 0.1
pKa = 0.2
```

```python
"""
Obtain the weak and strong acid

Input:
    c : acid 
Output:
    weak_acid : the result of the weak acid
    strong_acid : the result of the strong acid
"""
# Define the functions
def pH_of_strong_acid(c):
    return - log(c)

def pH_of_weak_acid(c, pKa):
    return -log(sqrt((c*pKa)))
    
```

```python
# =============================================================================
# Save the result and call the functions
# =============================================================================
strong_acid = pH_of_strong_acid(c)
weak_acid = pH_of_weak_acid(c, pKa)
```

```python
# =============================================================================
# Print the results
# =============================================================================
print('The strong acid solution is:', strong_acid)
print('The weak acid solution is:', weak_acid)
```

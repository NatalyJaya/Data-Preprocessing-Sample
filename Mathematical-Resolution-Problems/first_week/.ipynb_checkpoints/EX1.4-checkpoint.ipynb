- EX1.4 (1 point) = Write program for matrix multiplication using NumPy.
    - Note: This is the analogy of EX1.2 but solved automatically with NumPy.
    - Print the result for: `A = [[1,2],[3,4]]` and `B = [[5,6],[7,8]]`
    - Hint: Use standard matrix multiplication as shown in L02 - the problem is solved by one operator or function.

```python
import numpy as np
"""
Multiplies two matrices A and B using numpy.

Input:
    A : first matrix
    B : second matrix

Output:
    R : product of matrices A and B
"""

# Input matrix
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

# Calculate and save the result
R = A @ B

# Print the result
print('The result is:\n', R)
```

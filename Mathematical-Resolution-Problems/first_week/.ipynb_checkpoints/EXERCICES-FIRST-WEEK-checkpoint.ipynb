* EX1.1 (1 point) = Write a function to calculate the mass of a substance to get a solution of given concentration.
    - Inputs: conc. and volume of the solution, `c[mol/L]`,`V[L]`, and molar mass of your substance, `M[g/mol]`.
    - Output: mass of the substance, `m[g]`, you have to add to get the solution with required (`c`,`V`).
    - Requirement: add a docstring to your function (here: helpful; for complex functions: necessary).
    - Hint: remember the formulas from general chemistry (c = n/V; M = m/n).
    - Print the mass of NaBr to get 500 mL of the solution with concentration c = 0.2 mol/L (include units).

````python
"""
Calculates the mass of a substance needed to prepare a solution.

Input:
    c : concentration [mol/L]
    V : volume [L]
    M : molar mass [g/mol]

Output:
    m : mass of the substance [g]
"""

def calculate_mass(c, V, M):
    n = c * V
    m = M * n
    return m
````

````python
# Initialize the input parameters
c = 5
V = 4
M = 3

# Declare the result ant call the function to calculate
result = calculate_mass(c,V,M)

# Print the result
print(f"The mass with {c} mol/L, {V} L and {M} [g/mol] is {result} g")
````

* EX1.2 (1 point) = Write a short program for matrix multiplication.
    - Requirement: Use just 2D-lists and for-cycles, not numpy or other modules.
    - Hint: Copy+paste the code from Example 2 in Lecture 1 and modify it.
    - Print the result for: `A = [[1,2],[3,4]]` and `B = [[5,6],[7,8]]`

````python
"""
Multiplies two matrices A and B.

Input:
    A : first matrix
    B : second matrix

Output:
    result : product of matrices A and B
"""

# Input data
# A, B -> 2x2 matrix expressed as list
A = ([1,2],[3,4])
B = ([5,6],[7,8])

# Create an empty matrix to save the result
result = [[0 for _ in range(len(B[0]))] for _ in range(len(A))]

# Iterate through the rows of matrix A
for i in range(len(A)):

    # Iterate through the columns of matrix B
    for j in range(len(B[0])):

        # Calculate the element in position (i, j) into the result matrix
        for k in range(len(B)):
            result[i][j] += A[i][k] * B[k][j]

print(result)
````

````python
import math
import numpy as np
import matplotlib.pyplot as plt

"""
Generate and display the sine and cosine functions over a specified range.

Inputs:
    N_sample_points: Number of sample points used to generate the curves.
    x_min: Lower limit of the x-axis range.
    x_max: Upper limit of the x-axis range.

Output:
    A plot of the sine and cosine functions, saved as "ex1_5.png" and
    displayed on the screen.
"""

# Plotting variables
N_sample_points = 25
x_min = -2 * np.pi
x_max = 2 * np.pi

# Calculate the sine and cosine functions for x_values

# Create a list of N_sample_points from x_min to x_max
x_values = np.linspace(start=x_min,
                       stop=x_max,
                       num=N_sample_points,
                       endpoint=True)

y_sin = np.sin(x_values)
y_cos = np.cos(x_values)

# Add labels to the axes
plt.xlabel('x / rad')
plt.ylabel('f(x)')
plt.grid()

# Add data to the plot

# SINE
plt.plot(x_values,
         y_sin,
         color="green",
         linestyle="",
         marker='o',
         ms=3)

plt.plot(x_values,
         y_sin,
         label="sin(x)",
         color="red",
         linestyle="-",
         linewidth=2)

# COSINE
plt.scatter(x_values,
            y_cos,
            marker='^',
            color="red")

plt.plot(x_values,
         y_cos,
         label="cos(x)",
         color="blue",
         linestyle="--",
         linewidth=2)

plt.legend(loc="upper left",
           fontsize=11,
           markerscale=1.3)

# Limits of the axes
plt.xlim(-2 * np.pi, 2 * np.pi)
plt.ylim(-1.5, 1.5)

# Ticks on the x-axis
plt.xticks(
    [-2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2, 0,
     np.pi/2, np.pi, 3*np.pi/2, 2*np.pi],
    [r'$-2\pi$', r'$-3\pi/2$', r'$-\pi$', r'$-\pi/2$',
     r'$0$', r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$']
)

# Ticks on the y-axis
plt.yticks(np.arange(-1.5, 1.6, 0.5))

# Save the image
plt.savefig("ex1_5.png",
            dpi=300,
            bbox_inches='tight')

# Show the graph
plt.show()
````

````python
import numpy as np
import matplotlib.pyplot as plt

"""
Generate a comparison plot between experimental data and a reference function,
including uncertainty bands and residuals.

Inputs:
    N_sample_points: Number of sample points.
    x_min: Lower limit of the x-axis range.
    x_max: Upper limit of the x-axis range.
    figsize: Figure size.
    dpi: Figure resolution.
    x_label: Label for the x-axis.
    y_label: Label for the y-axis.

Output:
    A figure showing the reference function, experimental data with
    uncertainty, and the corresponding residuals. The figure is saved
    as "ex1_6.png" and displayed on the screen.
"""

# Data set
N_sample_points = 20  # Number of plotted points
x_min = 0             # Lower value of the data range
x_max = np.pi         # Upper value of the data range

figsize = (5,5)
dpi = 75
x_label = 'x / rad'
y_label = 'f(x)'

def ref_function(x):
    return np.cos(x) + np.cos(x**2)

x = np.linspace(x_min, x_max, N_sample_points)

y = np.array([ 2.14234667,  1.88911004,  1.79480494,  1.73942994,  1.43291172,  1.57731413,
          1.27047891,  0.98877325,  0.15915277, -0.52058165, -1.19948842, -1.39382868,
         -0.91996011, -0.63770412, -0.20389612,  0.0598823,  -0.38393458, -1.02237755,
         -2.13237608, -1.83003727])

error_y = y - ref_function(x)

# Create the figure and axes objects
fig = plt.figure(figsize=figsize, dpi=dpi)

# Main plot
ax1 = fig.add_axes([0, 0.5, 0.8, 0.5],
                   xticklabels=[]
                  )

# Residuals plot
ax2 = fig.add_axes([0, 0, 0.8, 0.5])

# Upper plot
ax1.plot(x,
         ref_function(x),
         '-',
         color='black',
         linewidth=1.5,
         label=r'$\cos(x) + \cos(x^2)$')

ax1.fill_between(x,
                 y - np.abs(error_y),
                 y + np.abs(error_y),
                 color='gray',
                 alpha=0.2,
                 label='Uncertainty')

ax1.errorbar(x,
             y,
             yerr=np.abs(error_y),       # Add vertical error bars
             fmt='s',                    # Use square markers
             markersize=3,
             color='tab:blue',
             label='Data')

# Lower plot: residuals
ax2.axhline(0,
            color='black',
            linestyle='--')

ax2.plot(x,
         error_y,
         's',
         color='blue')

# Axes limits
ax1.set_xlim(0,3)
ax2.set_xlim(0,3)

ax1.set_ylim(-2.5,2.5)
ax2.set_ylim(-0.5,0.5)

# Tick positions
ax2.set_xticks(np.arange(0,3.1,0.5))
ax1.set_yticks(np.arange(-2,3,1))
ax2.set_yticks([-0.3,0,0.3])

# Axis labels
ax1.set_ylabel(y_label, fontsize=16)
ax2.set_ylabel('Residuals', fontsize=16)
ax2.set_xlabel(x_label, fontsize=18)

# Grid
ax1.grid(True, linestyle='--', alpha=0.6)

# Apply styles (tick labels and border width) to both axes
for ax in [ax1, ax2]:
    ax.tick_params(labelsize=14)
    for spine in ax.spines.values():
        spine.set_linewidth(1.8)

ax1.legend(loc="upper right",
           fontsize=12,
           markerscale=0.9)

# Save the figure
fig.savefig('ex1_6.png', bbox_inches='tight')

# Display the figure
plt.show()
````

* EX 1.7 (2 points) = Make a template/script for the calculation of pH of a strong or a weak acid HA.
    - https://chem.libretexts.org/Bookshelves/Introductory_Chemistry &rarr; Search: pH of weak acid
    - Define two functions: pH_of_strong_acid(c), pH_of_weak_acid(c, pKa).
    - Print pH of a strong acid with c = 0.1, 0.2, 0.3, 0.4.

````python
from math import log, sqrt
"""
Obtain the weak and strong acid

Input:
    c : acid 
Output:
    weak_acid : the result of the weak acid
    strong_acid : the result of the strong acid
"""

# Input variables
c = 0.1
pKa = 0.2

# Define the functions
def pH_of_strong_acid(c):
    return - log(c)

def pH_of_weak_acid(c, pKa):
    return -log(sqrt((c*pKa)))
    
# Save the result and call the functions
strong_acid = pH_of_strong_acid(c)

# Print the results
weak_acid = pH_of_weak_acid(c, pKa)
print('The strong acid solution is:', strong_acid)
print('The weak acid solution is:', weak_acid)
    
````

* EX 1.8 (2 points) = Make a template/script for the calculation of interplanar spacing dhkl.
    - dhkl = interplanar spacing for an arbitrary crystallographic system and plane hkl
        - arbitrary system = the most general = triclinic = (a,b,c,alpha,beta,gamma)
        - hkl = diffraction indexes - see general chemistry or crystallography textbooks or Google
        - Wikipedia: https://en.wikipedia.org/wiki/Crystal_structure#Interplanar_spacing &rarr; Triclinic
    - Define a function of (nine! variables: `dhkl(a,b,c, alpha,beta,gamma, h,k,l)`.
    - Perform the calculation for (a,b,c, alpha,beta,gamma, h,k,l) = (4,5,6, 80,85,105, 1,1,1)
        - Note 1: The angles should be in degrees, but Python trigonometric functions work with radians.
        - Note 2: This is a formulla you definitely do NOT want to put in Excel (long, error-prone).


````python
import math

# Input variables
a = 4
b = 5
c = 6
alpha = 80
beta = 85
gamma = 105
h = 1
k = 1
l = 1

def dhkl(a, b, c, alpha, beta, gamma, h, k, l):
    """
    Calculate the interplanar spacing (d_hkl) for a triclinic crystal.

    Inputs:
        a, b, c: Unit cell lengths.
        alpha, beta, gamma: Unit cell angles in degrees.
        h, k, l: Miller indices.

    Output:
        result: Interplanar spacing (d_hkl).
    """

    # Convert the cell angles from degrees to radians
    r_alpha = math.radians(alpha)
    r_betha = math.radians(beta)
    r_gamma = math.radians(gamma)

    # Compute the sine and cosine of each angle
    s_alpha = math.sin(r_alpha)
    s_betha = math.sin(r_betha)
    s_gamma = math.sin(r_gamma)

    c_alpha = math.cos(r_alpha)
    c_betha = math.cos(r_betha)
    c_gamma = math.cos(r_gamma)

    # Calculate each term of the numerator separately for better readability
    t1 = ((h**2) / (a**2)) * (s_alpha**2)
    t2 = ((k**2) / (b**2)) * (s_betha**2)
    t3 = ((l**2) / (c**2)) * (s_gamma**2)

    t4 = ((2 * k * l) / (b * c)) * (c_betha * c_gamma - c_alpha)
    t5 = ((2 * h * l) / (a * c)) * (c_gamma * c_alpha - c_betha)
    t6 = ((2 * h * k) / (a * b)) * (c_alpha * c_betha - c_gamma)

    # Compute the numerator
    numerator = t1 + t2 + t3 + t4 + t5 + t6

    # Compute the denominator
    denominator = (
        1
        - c_alpha**2
        - c_betha**2
        - c_gamma**2
        + 2 * c_alpha * c_betha * c_gamma
    )

    # Calculate the inverse squared spacing
    division = numerator / denominator

    # Compute the interplanar spacing
    result = math.sqrt(1 / division)

    return result

# Declare the result of the function
result = dhkl(a, b, c, alpha, beta, gamma, h, k, l)

# Print the result
print("The interplanar spacing is:", result)
````

- EX1.9 (1 point) = Conversion among degees of Celsius, Kelvin, Farhenheit, and Reaumur using NumPy.
    - Print the table for the following temperatures: -20, 0, 20, 40, 60, 80, 100 C.
    - Hint: Use NumPy vector computing to simplify the calculations - you will need no for-cycles.

````python
import numpy as np

# Temperature values in degrees Celsius
celsius = np.array([-20, 0, 20, 40, 60, 80, 100])

"""
Convert temperatures from Celsius to Kelvin, Fahrenheit, and Réaumur.

Inputs:
    celsius: array containing temperatures in degrees Celsius.

Outputs:
    tabla_temperaturas: array where each row contains the temperature
    in Celsius, Kelvin, Fahrenheit, and Réaumur, respectively.
"""

# Convert Celsius to Kelvin
kelvin = celsius + 273.15

# Convert Celsius to Fahrenheit
fahrenheit = (celsius * 9 / 5) + 32

# Convert Celsius to Réaumur
reaumur = celsius * 4 / 5

# Combine all temperature scales into a single table
temperatures_table = np.array([celsius, kelvin, fahrenheit, reaumur]).T

# Display the resulting temperature conversion table
print(temperatures_table)
````

````python
import numpy as np
import matplotlib.pyplot as plt

"""
Generate and display contour and filled contour plots of the function
z = cos(x) + sin(y).

Inputs:
    N_sample_points: Number of sample points.
    x_min: Lower limit of the x range.
    x_max: Upper limit of the x range.
Output:
    A contour plot and a filled contour plot with a color bar representing
    the values of z.
"""

N_sample_points = 20          # Number of points plotted
x_min = 0                     # Lower value of the range of the data
x_max = 3 * np.pi             # Higher value of the range of the data

# Creates a list of N_sample_points from x_min to x_max
x_values = np.linspace(start=x_min,
                       stop=x_max,
                       num=N_sample_points)

# Create two 2D coordinate grids from the x values
x_2darray, y_2darray = np.meshgrid(x_values, x_values)

# Evaluate the function z = cos(x) + sin(y) on the grid
z_2darray = np.cos(x_2darray) + np.sin(y_2darray)

# Draw contour lines
plt.contour(x_2darray, y_2darray, z_2darray, 20, cmap='RdBu')

# Draw filled contours
plt.contourf(x_2darray, y_2darray, z_2darray, 20, cmap='RdBu')

# Add a color bar to indicate the magnitude of z
magnitude = plt.colorbar(ticks=[-2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2])
magnitude.set_label('Z')

# Label the axes
plt.xlabel('x / rad')
plt.ylabel('y / rad')

# Display the figure
plt.show()
````

- EX1.12 Create a Python script to produce the following plot using Matplotlib. Requirements:
    - Produce both a 2D histogram and a contour overlay
    - Print the covariance matrix to the console
    - Save as `ex1.12.png`
    - HINT: the data can be generated using the following lines of code:
    
```python
    rng = np.random.default_rng(0)
    mean = [0, 0]
    cov = [[1, 1], [1, 2]]
    x, y = np.random.multivariate_normal(mean, cov, 3000).T
```

Solution:

<img src="figs/ex3_A.png" alt="image" width="400" height="auto">


````python
# =============================================================================
# Import modules
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt
````

````python
"""
Generate a scatter plot of a bivariate normal distribution together with
its marginal histograms.

Inputs:
    nrows: Number of subplot rows.
    ncols: Number of subplot columns.
    mean: Mean vector of the bivariate normal distribution.
    cov: Covariance matrix of the distribution.

Output:
    A figure containing the scatter plot of the generated samples and
    the corresponding marginal histograms.
"""
# (1) Random number generator
rng = np.random.default_rng(0)
# (2) Mean vector and covariance matrix
mean = [0, 0]
cov = [[1, 1], [1, 2]]
# (3) Generate random samples from a bivariate normal distribution
x, y = np.random.multivariate_normal(mean, cov, 3000).T

# =============================================================================
# Display the covariance matrix
# =============================================================================
print("Covariance Matrix:")
print(cov)

# =============================================================================
# Create the figure
# =============================================================================
fig = plt.figure(figsize=(6,6))

grid = plt.GridSpec(2,            # Number of rows
                    2,            # Number of columns
                    wspace=0.05,  # Horizontal space between subplots
                    hspace=0.05,  # Vertical space between subplots
                    width_ratios=(1, 4),      # Narrow left panel, wide right panel
                    height_ratios=(4, 1),     # Wide top panel, narrow bottom panel
                   )

# =============================================================================
# Main scatter plot
# =============================================================================
ax_hist2d = fig.add_subplot(grid[0, 1])
ax_hist2d.scatter(x,
                  y,
                  color='green',
                  alpha=0.3,
                  s=10)
ax_hist2d.tick_params(labelbottom=False, labelleft=False)

# =============================================================================
# Left subplot: histogram of y values
# =============================================================================
ax_histy = fig.add_subplot(grid[0, 0], sharey=ax_hist2d)
ax_histy.hist(y,
              bins=30,
              density=True,
              orientation='horizontal',
              color='tab:blue')
ax_histy.invert_xaxis()
ax_histy.tick_params(labelbottom=False)
ax_histy.set_yticks([-5.0, -2.5, 0.0, 2.5, 5.0])

# =============================================================================
# Bottom subplot: histogram of x values
# =============================================================================
ax_histx = fig.add_subplot(grid[1, 1], sharex=ax_hist2d)
ax_histx.hist(x,
              bins=30,
              density=True,
              color='tab:blue')
ax_histx.tick_params(labelleft=False)
ax_histx.set_xticks([-2.5, 0.0, 2.5])

# =============================================================================
# Save the required image
# =============================================================================
plt.savefig('ex1.12.png', bbox_inches='tight', dpi=300)

# =============================================================================
# Display the figure
# =============================================================================
plt.show()
````

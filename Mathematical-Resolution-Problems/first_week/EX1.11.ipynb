````python
# =============================================================================
# Import modules
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt
````

````python
"""
Generate and display contour and filled contour plots of the function
z = cos(x) + sin(y).

Inputs:
    N_sample_points: Number of sample points.
    x_min: Lower limit of the x range.
    x_max: Upper limit of the x range.
Output:
    A contour plot and a filled contour plot with a color bar representing
    the values of z.
"""
# =============================================================================
# N_sample_points and range limits
# =============================================================================
N_sample_points = 20          # Number of points plotted
x_min = 0                     # Lower value of the range of the data
x_max = 3 * np.pi             # Higher value of the range of the data

# =============================================================================
# Creates a list of N_sample_points from x_min to x_max
# =============================================================================
x_values = np.linspace(start=x_min,
                       stop=x_max,
                       num=N_sample_points)

# =============================================================================
# Create two 2D coordinate grids from the x values
# =============================================================================
x_2darray, y_2darray = np.meshgrid(x_values, x_values)

# =============================================================================
# Evaluate the function z = cos(x) + sin(y) on the grid
# =============================================================================
z_2darray = np.cos(x_2darray) + np.sin(y_2darray)

# =============================================================================
# Draw contour lines
# =============================================================================
plt.contour(x_2darray, y_2darray, z_2darray, 20, cmap='RdBu')

# =============================================================================
# Draw filled contours
# =============================================================================
plt.contourf(x_2darray, y_2darray, z_2darray, 20, cmap='RdBu')

# =============================================================================
# Add a color bar to indicate the magnitude of z
# =============================================================================
magnitude = plt.colorbar(ticks=[-2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2])
magnitude.set_label('Z')

# =============================================================================
# Label the axes
# =============================================================================
plt.xlabel('x / rad')
plt.ylabel('y / rad')

# =============================================================================
# Display the figure
# =============================================================================
plt.show()
````

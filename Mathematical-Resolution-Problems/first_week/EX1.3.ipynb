- EX1.3 (1 point) = Solve quadratic equation: `a*x**2 + b*x + c = 0` for `(a,b,c) = (1,1,-1)` using NumPy.
    - Note: This is an analogy of the sample program from L01 (DEMOS, notebook 2quadratic_eq).
    - In standard Python, we needed a whole program to solve quadratic equation, while in NumPy - ONE command!
    - Hint: Use numpy.roots to find the solutions - this is the one (and only) command you need.

```python
import numpy as np

"""
Calculates the roots of a quadratic equation using NumPy.

Input:
    a : first coefficient
    b : second coefficient
    c : third coefficient

Output:
    result : roots of the quadratic equation
"""

# (1) Input data
a = 1
b = 1
c = -1
```

```python
# (2) Calculate the roots of the polynomial ax² + bx + c
result = np.roots([a, b, c])
```

```python
# (3)Print the results
print("The roots are:", result)
```

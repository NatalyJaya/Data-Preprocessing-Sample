- EX1.10 (1 point) = Solve cubic equation: `a*x**3 + b*x**2 + c*x + d = 0` for `(a,b,c,d) = (1,-6, 11,-6)` using NumPy.
    - Note: This an analogy of EX1.B that can be automatically solved with NumPy.

```python
# =============================================================================
# Import modules
# =============================================================================
import numpy as np
```

```python
# =============================================================================
# Coefficients of the cubic polynomial
# =============================================================================
a = 1
b = -6
c = 11
d = -6
```

```python

"""
Calculate the roots of a cubic polynomial.

Inputs:
    a, b, c, d: Coefficients of the polynomial
                ax³ + bx² + cx + d = 0.

Output:
    result: NumPy array containing the roots of the polynomial.
"""

# (1) Compute the roots of the polynomial
result = np.roots([a, b, c, d])
```

```python
# =============================================================================
# Display the roots
# =============================================================================
print("The roots are: ", result)
```

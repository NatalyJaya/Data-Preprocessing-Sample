## Motivation for the exercises

* We should practice two  things:
    - How to define and use functions in Python - common task in research.
    - How to write simple algorithms in Python - important basic knowledge.
* Why should we know how to define/use our functions?
    - Almost everything in research is a function ...
    - Just a few examples: pH, solubility of a polymer, diffraction intensity ...
* Why should we be able to write/understand basic algorithms? 
    - In scientific computing, we usually use optimized algorithms from various libraries.
    - BUT some idea how to write our own small programs will help us solve real-life problems.
* Some tasks are the same like but ask you to solve the exercises "manually" or "automatically" with NumPy.  This is an important demonstration of several important facts:
    - Many common task can be solved with Python libraries (such as NumPy).
    - The solutions based on prepared library functions are *always* easier - even for a beginner.
    - Moreover, the algorithms in Python libraries are safe and ultrafast - prefer them whenever possible.

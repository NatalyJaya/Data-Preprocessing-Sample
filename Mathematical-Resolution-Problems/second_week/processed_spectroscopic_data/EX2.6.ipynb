- EX2.6 (1 point) Do a program that provides the solution to the following system of linear equations

$\left\{\begin{array}{@{}l@{}}
    x_1 + 9x_2 +2x_3 +x_4 +x_5 = 170\\
    10x_1 + x_2 +2x_3 +x_4 +x_5 = 180\\
    x_1  +5x_3 +x_4 +x_5 = 140\\
    2x_1 +  x_3 +2x_4 +9x_5 = 180\\
    2x_1 + x_2 +2x_3 +13x_4 +2x_5 = 350
  \end{array}\right.\,$

```python
import numpy as np
from scipy.linalg import solve

"""
Solve a system of linear equations in matrix form Ax = b.

Inputs:
    A: Coefficient matrix of the linear system.
    b: Right-hand-side vector containing the constant terms.

Output:
    x: NumPy array containing the solution of the linear system.
"""

# Define the coefficient matrix
A = np.array([
    [1, 9, 2, 1, 1],
    [10, 1, 2, 1, 1],
    [1, 0, 5, 1, 1],
    [2, 0, 1, 2, 9],
    [2, 1, 2, 13, 2]
])

# Define the right-hand-side vector
b = np.array([170, 180, 140, 180, 350])

# Solve the linear system Ax = b
solution = solve(A, b)

# Display the solution vector
print('The solution:', solution)
```

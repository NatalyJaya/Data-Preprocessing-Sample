```python
import seaborn as sns
import matplotlib.pyplot as plt

"""
Load the penguins dataset and generate a joint plot showing the relationship
between bill length and bill depth, grouped by species.

Inputs:
    None (the penguins dataset is loaded directly from Seaborn).

Output:
    A joint plot of bill length versus bill depth, colored by species,
    saved as "ex52-3.nb.png" and displayed on the screen.
"""

# Load the penguins dataset into a DataFrame
df = sns.load_dataset('penguins')

# Configure the global plotting style and font size
sns.set_theme(style='darkgrid', font_scale=1.2)

# Create the joint plot
fig = sns.jointplot(
    data=df,                 # DataFrame containing the data
    x="bill_length_mm",      # Variable displayed on the x-axis
    y="bill_depth_mm",       # Variable displayed on the y-axis
    hue='species',           # Color the data according to penguin species
    palette='Reds'           # Use the Reds color palette
)

# Set the axis labels
fig.ax_joint.set_xlabel('Bill length (mm)')
fig.ax_joint.set_ylabel('Bill depth (nm)')

# Save the figure with high resolution
fig.fig.savefig('ex52-3.nb.png', dpi=300)

# Display the figure
plt.show()
```

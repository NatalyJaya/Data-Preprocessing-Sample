* Read the testing database from the main lecture L05 (`DATA\uhmwpe_thr.xlsx`) to a dataframe named `df`.
* Using column selection and boolean indexing, create the following sub-database and save it as a dataframe `df2`.
    - Columns: only 'ID', 'ImplantName', 'Sterilization', 'Dose_kGy', 'IR/OI', 'IR/VI'
    - Rows: only crosslinked samples (Dose_kGy > 0) that were EtO-sterilized (Sterilization = 'EtO').
* Sort the subdatabase df2 according to values of vinylene index ('IR/VI') and print it.
* In the final comment, answer the following question:
    - Is there some correlation between radiation dose and VI?

```python
import pandas as pd

"""
Load an Excel database, filter the data according to the selected criteria,
sort the results by the vinyl index, and display the filtered dataset.

Inputs:
    XLSX_FILE: Path to the Excel file containing the experimental data.

Output:
    A filtered and sorted pandas DataFrame containing the selected columns,
    printed to the console.
"""

# Load the experimental database from the Excel file
XLSX_FILE = 'DATA/uhmwpe_thr.xlsx'
df = pd.read_excel(XLSX_FILE,
                   sheet_name='DataBEFOREaging',
                   skiprows=17,
                   skipfooter=11)

# Apply boolean indexing and select the requested columns
filtered = (df['Dose_kGy'] > 0) & (df['Sterilization'] == 'EtO')

df2 = df.loc[
    filtered,
    ['ID', 'ImplantName', 'Sterilization', 'Dose_kGy', 'IR/OI', 'IR/VI']
]

# Sort the filtered DataFrame by the vinyl index (IR/VI)
df2_sorted = df2.sort_values(by='IR/VI', ascending=True)

# Display the sorted DataFrame
print(df2_sorted)
```

### Is there a correlation between the radiation dose and the vinyl index (VI)?
As the absorbed radiation dose (Dose_kGy) increases,the vinyl index (IR/VI) also increases.

- EX2.6 (1 point) Do a program that provides the solution to the following system of linear equations

$\left\{\begin{array}{@{}l@{}}
    x_1 + 9x_2 +2x_3 +x_4 +x_5 = 170\\
    10x_1 + x_2 +2x_3 +x_4 +x_5 = 180\\
    x_1  +5x_3 +x_4 +x_5 = 140\\
    2x_1 +  x_3 +2x_4 +9x_5 = 180\\
    2x_1 + x_2 +2x_3 +13x_4 +2x_5 = 350
  \end{array}\right.\,$

```python
import numpy as np
from scipy import linalg

"""
Solve a system of five linear equations using matrix algebra.

Inputs:
    A: Coefficient matrix of the linear system.
    b: Right-hand-side vector containing the constant terms.

Output:
    The values of the unknowns x1, x2, x3, x4, and x5, printed to the console.
"""

# Define the coefficient matrix of the linear system
A = np.array(
    [
        [1, 9, 2, 1, 1],
        [10, 1, 2, 1, 1],
        [1, 0, 5, 1, 1],
        [2, 0, 1, 2, 9],
        [2, 1, 2, 13, 2],
    ]
)

# Define the right-hand-side vector
b = np.array([170, 180, 140, 180, 350])

# Solve the linear system A · x = b
x = linalg.solve(A, b)

# Unpack the solution vector into individual variables
x1, x2, x3, x4, x5 = x

# Display the solution
print(f"x1={x1}, x2={x2}, x3={x3}, x4={x4}, x5={x5}")
```
